```python
#Rectangle Class:Calculate Area,Perimeter and Compare Two Rectangles.
class rectangle:
    def __init__(self,l,b):
        self.l=l
        self.b=b
    def area(self,l,b):
        return l*b
    def perimeter(self,l,b):
        return 2*(l+b)
l1=int(input("Enter length of Rectangle1:"))
print("length of rectangle1:",l1)
b1=int(input("Enter breadth of Rectangle1:"))
print("breadth of rectangle1:",b1)
l2=int(input("Enter length of Rectangle2:"))
print("length of rectangle2:",l2)
b2=int(input("Enter breadth of Rectangle2:"))
print("breadth of rectangle2:",b2)
r1=rectangle(l1,b1)
r2=rectangle(l2,b2)
p=r1.area(l1,b1)
q=r2.area(l2,b2)
print("Area of Rectangle1:",p)
print("Area of Rectangle2:",q)
print("Perimeter of Rectangle1:",r1.perimeter(l1,b1))
print("Perimeter of Rectangle2:",r2.perimeter(l2,b2))
if(p>q):
    print("Rectangle 1 is greater")
elif(p<q):
    print("Rectangle 2 is greater")
else:
    print("Rectangles are equal")
```

```python
#Bank Account Class:Deposit,Withdraw and Display Account Details.
class bankaccount:
    def __init__(self,a_name,a_no,a_type,a_balance):
        self.acc_name=a_name
        self.acc_no=a_no
        self.acc_type=a_type
        self.acc_balance=a_balance
    def deposit(self,a_deposit):
        print("Initial balance:",self.acc_balance)
        self.acc_balance+=a_deposit
        print(a_deposit,"is deposited")
        print("current balance: ",self.acc_balance)
    def withdraw(self):
        self.amount=int(input("Withdraw amount:"))
        print("Initial balance:",self.acc_balance)
        if self.amount > self.acc_balance:
            print("You don't have enough balance to withdraw !!")
            print("Current balance: ",self.acc_balance)
        else:
            print(self.amount,"is withdrawn")
            self.acc_balance-=self.amount
            print("Current balance: ",self.acc_balance)
    def account_info(self):
        print("Account holder Name:",self.acc_name)
        print("Account Number:",self.acc_no)
        print("Account Type:",self.acc_type)
        print("Account Balance:",self.acc_balance)
name=input("Enter name:")
no=int(input("Enter account number:"))
type=input("Enter account type:")
balance=int(input("Enter initial balance:"))
b1=bankaccount(name,no,type,balance)
while True:
    print("1. Deposit 2. Withdraw 3. Account info 4. Exit")
    ch=int(input("Choose your option:"))
    if ch==1:
        d=int(input("Deposit amount:"))
        b1.deposit(d)
    elif ch==2:
        b1.withdraw()
    elif ch==3:
        b1.account_info()
    elif ch==4:
        print("Thank you")
        break
    else:
      print("Invalid choice")


```

```python
#Multilevel Inheritance:-Python Book Details.
class publisher:
    def __init__(self,pname):
        self.pubname=pname
    def display1(self):
        print("Book details")
        print("Publisher name=",self.pubname)
class book(publisher):
    def __init__(self,pname,bname,author):
        self.title=bname
        self.auth=author
        publisher.__init__(self,pname)
    def display2(self):
        print("Title=",self.title)
        print("Author",self.auth)
class python(book):
    def __init__(self,pname,title,au,price,np):
        self.rate=price
        self.nop=np
        book.__init__(self,pname,title,au)
    def display3(self):
        print("Rate=",self.rate)
        print("No. of pages=",self.nop)
p=input("Enter publisher name:")
b=input("Enter book name:")
a=input("Enter author name: ")
r=int(input("Enter rate of book:"))
n=int(input("Enter no. of pages:"))
a=python(p,b,a,r,n)
a.display1()
a.display2()
a.display3()

```

```python
#Operator Overloading:-Compare Rectangle Areas
class rectangle:
    def __init__(self,l,w):
        self.length=l
        self.width=w
        self.a=l*w
    def __lt__(self,other):
        if self.a < other.a:
            print("Rectangle 2 is greater")
        else:
            print("Rectangle 1 is greater")
l1=int(input("Enter length of rectangle1:"))
w1=int(input("Enter width of rectangle1:"))
print("Length of Rectangle1:",l1)
print("Width of Rectangle1:",w1)
l2=int(input("Enter length of rectangle2:"))
w2=int(input("Enter width of rectangle2:"))
print("Length of Rectangle2:",l2)
print("Width of Rectangle2:",w2)
r1=rectangle(l1,w1)
r2=rectangle(l2,w2)
r1<r2
```

```python
#Operator Overloading:-Adiition of Two Time Object.
class Time:
    def __init__(self,hr,mint,sec):
        self.__h=hr
        self.__m=mint
        self.__s=sec
    def __add__(self,other):
        hh=self.__h+other.__h
        mm=self.__m+other.__m
        ss=self.__s+other.__s
        if ss>=60:
            m=int(ss/60)
            ss=ss%60
            mm+=m
        if mm>=60:
            h=int(mm/60)
            mm=mm%60
            hh+=h
        print("Added time..")
        print("Hour:",hh,"Minute:",mm,"Second:",ss)

h1=int(input("Enter hour1:"))
m1=int(input("Enter minute1:"))
s1=int(input("Enter second1:"))
print("Hour1:",h1)
print("Minute1:",m1)
print("Second1:",s1)
h2=int(input("Enter hour2:"))
m2=int(input("Enter minute2:"))
s2=int(input("Enter second2:"))
print("Hour2:",h2)
print("Minute2:",m2)
print("Second2:",s2)
t1=Time(h1,m1,s1)
t2=Time(h2,m2,s2)
t1+t2
```
