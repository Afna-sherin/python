```python
#list
numbers = list(map(int, input("Enter the numbers: ").split()))
word = input("Enter a word:") 
print("original integers:",numbers)
print("original word is:",word)
positive_numbers = [num for num in numbers if num > 0]
square=[x**2 for x in numbers]
vowels=[ch for ch in word if ch.lower() in 'aeiou']
ordinal_values=[ord(ch) for ch in word]
print("positive numbers:",positive_numbers)
print("square of numbers:",square)
print("Vowels:", vowels)
print("Ordinal values:", ordinal_values)

```
