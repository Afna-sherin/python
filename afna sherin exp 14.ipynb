```python
%%writefile graphics/rectangle.py
#Rectangle module.
def area(l,b):
    return l*b
def perimeter(l,b):
    return 2*(l+b)
    
```

```python
 %%writefile graphics/circle.py
#Circle module.
import math
def area(r):
    return math.pi*r*r
def perimeter(r):
    return 2*math.pi*r
```

```python
%%writefile graphics/graphics3D/cuboid.py
#Cube module.
def area(l,b,h):
    return 2*(l*b+l*h+b*h)
def perimeter(l,b,h):
    return 4*(l+b+h)
```
