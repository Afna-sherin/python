```python
#change a given string
s=input("Enter a string:")
new=s[-1]+s[1:-1]+s[0]
print("new string:",new)
```

```python
#first 2 and the last 2 characters
s=input("Enter a string:")
new=s[:2]+s[-2:]
print("new string:",new)
```

```python
#single string seperated with space
s1=input("Enter a first string:")
s2=input("Enter a second string:")
a=s2[0]+s1[1:]
b=s1[0]+s2[1:]
print(a , b)
```

```python
#replace a string
s=input("enter a string:")
ns=s[0]+s[1:].replace(s[0],"$")
print("new string :",ns)
```
