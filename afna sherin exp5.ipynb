```python
tuples=tuple(map(int,input("enter the element:").split(",")))
print("tuple is:",tuples)
print("3rd element is:",tuples[2])
print("3rd element from the last is:",tuples[-3])
```

```python
#average of all items
numbers=input("enter the numbers:")
tuples=tuple(map(int,numbers.split(",")))
print("tuple:",tuples)
print("average:",sum(tuples)/len(tuples))
```
