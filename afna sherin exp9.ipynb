```python
#string
s=input("Enter a string:")
if s[-3:]=="ing":
    string=s+"ly"
else:
    string=s+"ing"
print("original string is:",s)
print("new string:",string)
```

```python
#palindrome
s=input("Enter a string:")
print("original string is:",s)
print("reversed string is:",s[::-1])
if s==s[::-1]:
    print("it is a palindrome")
else:
    print("it is not palindrome")
```
